```python
print("Q3. Relationship between screen time and sleep hours")
print(df[['screen_time', 'sleep_hours']].corr())

plt.figure(figsize=(8, 6))
plt.scatter(df['screen_time'], df['sleep_hours'], c='blue', s=50, alpha=0.5)
plt.xlabel('Screen Time (hours)')
plt.ylabel('Sleep Hours')
plt.title('Screen Time vs Sleep Hours')
plt.grid(True, alpha=0.3)
plt.savefig('q3_screen_vs_sleep.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
data = df['daily_transport_cost']
plt.figure(figsize=(10, 6))
plt.hist(data, bins=20, color='blue', edgecolor='black', alpha=0.7)
plt.xlabel('Daily Transport Cost')
plt.ylabel('Frequency')
plt.title('Histogram of Daily Transport Cost')
plt.savefig('q4_transport_cost_hist.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
internet_by_gender = df.groupby('gender')['internet_usage'].mean()
print("Q5. Average internet usage by gender:")
print(internet_by_gender)
```

# Females on average used more internet than males

```python
plt.figure(figsize=(6, 5))
plt.bar(internet_by_gender.index, internet_by_gender.values, color='green', width=0.5)
plt.xlabel('Gender')
plt.ylabel('Average Internet Usage (hours)')
plt.title('Average Internet Usage by Gender')
plt.savefig('q5_internet_by_gender.png', dpi=300, bbox_inches='tight')
plt.show()
```
