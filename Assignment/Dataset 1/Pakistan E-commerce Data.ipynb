```python
city_counts = df.groupby("city")["order_value"].count().sort_values(ascending=False)
print(f"City with the highest number of orders: {city_counts}")
```

```python
pay_counts= df.groupby("payment_method")["order_value"].count().sort_values(ascending=False)
print(f"Most popular payment method: {pay_counts}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
avg_by_city = df.groupby("city")["order_value"].mean()
categories = avg_by_city.index
values = avg_by_city.values

plt.figure(figsize=(10, 6))                  
plt.bar(categories, values, color='green', width=0.5)
plt.xlabel('City')
plt.ylabel('Average Order Value')
plt.title('Average Order Value by City')                      
plt.grid(True, alpha=0.3)
plt.savefig('q5_avg_order_by_city.png', dpi=300, bbox_inches='tight')
plt.show()
```
