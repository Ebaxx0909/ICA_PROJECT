```python
avg_fare_per_service = df.groupby('service_type')['fare'].mean()
print(f"Average fare per service type: \n{avg_fare_per_service}")
```

```python
city_highest_trips = df.groupby('city')['trip_id'].count().idxmax()
print(f"City with the highest number of trips: {city_highest_trips}")
```

```python
avg_distance_km = df['distance_km'].mean()
print(f"Average distance traveled per trip: {avg_distance_km} km")
```

```python
Trip_rating_by_gender = df.groupby('gender')['trip_rating'].mean()
print(f"Average trip rating by gender: \n{Trip_rating_by_gender}")
```

# On average male customers give higher ratings to riders in comparision to women

```python
plt.figure(figsize=(6, 5))
plt.bar(Trip_rating_by_gender.index, Trip_rating_by_gender.values, color='green', width=0.5)
plt.xlabel('Gender')
plt.ylabel('Average Trip Rating')
plt.title('Average Trip Rating by Gender')
plt.savefig('q5_trip_rating_by_gender.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(6, 5))
plt.bar(avg_fare_per_service.index, avg_fare_per_service.values, color='green', width=0.5)
plt.xlabel('service_type')
plt.ylabel('avg_fare (Rs)')
plt.title('Average Fare per Service Type')
plt.savefig('q5_avg_fare_per_service.png', dpi=300, bbox_inches='tight')
plt.show()
```
