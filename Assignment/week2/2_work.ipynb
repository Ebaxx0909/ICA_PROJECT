```python
x["price"] = x["price"].str.replace(",", "", regex=False)
```

```python
# Rows above the IQR upper fence
q1, q3 = x["price"].quantile([0.25, 0.75])
iqr = q3 - q1
upper = q3 + 1.5 * iqr
lower = q1 - 1.5 * iqr

print(x[(x["price"] > upper) | (x["price"] < lower)])
```

```python
x["price"] = x["price"].replace(9999, pd.NA)
x["price"] = pd.to_numeric(x["price"], errors="coerce")

# Fill all missing prices (the original NaN + the 9999s) with the median
x["price"] = x["price"].fillna(x["price"].median())

# Or drop them instead:
# x = x.dropna(subset=["price"])
```
